```python
def avaliar(x):
  return 2*x
```

```python
def expandir(x):
  vizinhos = []
  if x > 0:
    vizinhos.append(x-1)
  if x < 10:
    vizinhos.append(x+1)
  return vizinhos
```

```python
def subida_de_encosta(estado_inicial,avaliar,expandir):
  atual=estado_inicial

  while True:
    print(f"Estado: {atual} Avaliação: {avaliar(atual)}")

    vizinhos=expandir(atual)

    if not vizinhos:
      return atual

    melhor_vizinho = max(vizinhos,key=avaliar)

    if avaliar(melhor_vizinho) <= avaliar(atual):
      return atual

    atual = melhor_vizinho


```

```python
resultado = subida_de_encosta(1,avaliar,expandir)
print(resultado)
```
